```python
# Define the labyrinth (2D Array with emojis)
labyrinth = [
    ["🧱", "🧱", "🧱", "🧱", "🧱", "🧱", "🧱", "🧱", "🧱"],
    ["🧱", "🧍", "⬜", "🧱", "⬜", "⬜", "⬜", "⬜", "🧱"],
    ["🧱", "⬜", "🧱", "🧱", "⬜", "🧱", "🧱", "⬜", "🧱"],
    ["🧱", "⬜", "⬜", "⬜", "⬜", "⬜", "🧱", "⬜", "🧱"],
    ["🧱", "🧱", "⬜", "🧱", "🧱", "⬜", "🧱", "⬜", "🧱"],
    ["🧱", "⬜", "⬜", "⬜", "⬜", "⬜", "⬜", "⬜", "🧱"],
    ["🧱", "🧱", "🧱", "🧱", "🧱", "🧱", "🧱", "🚪", "🧱"]
]

# Initial position of the player
pos_x, pos_y = 1, 1

# A function to display the labyrinth
def display_labyrinth():

    print("\n" * 10) # Simulate output clearance

    for row in labyrinth:
      for element in row:
        print(element, end=" ")
      print()
    print("\n🔹 Objective: Find the exit (🚪)!")

# Game loop
leave = False

while leave == False:
    display_labyrinth()

    move = input("Move (WASD): ").upper()

    # Calculate the new position
    new_x, new_y = pos_x, pos_y
    if move == "W":  # Up
        new_x -= 1
    elif move == "S":  # Down
        new_x += 1
    elif move == "A":  # Left
        new_y -= 1
    elif move == "D":  # Right
        new_y += 1
    else:
        print("❌ Input is not valid! Please use W, A, S, or D instead.")
        continue

    # Check whether the move is valid
    if not (0 < new_x < len(labyrinth) and
            0 < new_y < len(labyrinth[0]) and
            labyrinth[new_x][new_y] != "🧱"):

        print("🚧 You hit a wall!")
        continue

    # Check whether the player have reached the exit
    if labyrinth[new_x][new_y] == "🚪":
        display_labyrinth()
        print("🎉 Congratulations! You have successfully leave the labyrinth!")
        leave = True

    # Update the player position
    labyrinth[pos_x][pos_y] = "⬜"  # Mark the last position as open
    pos_x, pos_y = new_x, new_y # Change the last position with the new one
    labyrinth[pos_x][pos_y] = "🧍"  # Update the new player position
```
